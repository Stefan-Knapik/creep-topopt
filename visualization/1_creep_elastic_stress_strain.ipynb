# Stress and strain ranges of the creep and elastic designs (Sec. 4.1)

Prints the equivalent stress and creep strain ranges of each design, taken over the cells whose filtered (physical) density is above 0.5. The filtered density is the field the solver analysed, so these are the material cells of the simulated structure. The paper quotes the stress ranges in Sec. 4.1. This notebook generates no figure.

The fields come from `results/1_creep_elastic/data_evaluation/reconstructed_fields.npz`, written by `experiments/1_creep_elastic/reconstruct_stress_strain_fields.py`. It holds the same creep simulation for both designs, which the elastic design's `.vtu` doesn't have.

```python
import numpy as np

from plot_utils import RESULTS, equivalent_strain, von_mises_stress

fields = np.load(RESULTS / "1_creep_elastic" / "data_evaluation" / "reconstructed_fields.npz")

for tag in ("elastic_50", "creep_50"):
    # Range over cells with filtered (physical) density > 0.5, not over the plotting threshold.
    solid = fields[f"{tag}__rho"] > 0.5
    sig = von_mises_stress(fields[f"{tag}__stress_0"])[solid]
    eps = equivalent_strain(fields[f"{tag}__creep_strain_final"])[solid]
    print(f"{tag:<12s} cells with filtered density > 0.5: {solid.sum()} of {solid.size}")
    print(f"{tag:<12s} initial sigma_eq: {sig.min():.4f} to {sig.max():.4f} MPa")
    print(f"{tag:<12s} final eps_eq:     {eps.min():.6e} to {eps.max():.6e}")
```
