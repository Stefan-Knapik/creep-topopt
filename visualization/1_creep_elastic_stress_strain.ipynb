# Stress and strain ranges of the creep and elastic designs (Sec. 4.1)

Prints the equivalent stress and creep strain ranges over the solid cells of each design. The paper quotes the stress ranges in Sec. 4.1. This notebook generates no figure.

The fields come from `results/1_creep_elastic/data_evaluation/reconstructed_fields.npz`, written by `experiments/1_creep_elastic/reconstruct_stress_strain_fields.py`. It holds the same creep simulation for both designs, which the elastic design's `.vtu` doesn't have.

```python
import numpy as np

from plot_utils import RESULTS, equivalent_strain, load_design, solid_bounds, von_mises_stress

VF = 0.5
RUN_DIR = RESULTS / "1_creep_elastic"
DESIGNS = {
    "elastic_50": RUN_DIR / "data_elastic_50" / "vtk" / "sol_250.vtu",
    "creep_50": RESULTS / "2_pareto_vf" / "data_creep_50" / "vtk" / "sol_250.vtu",
}

fields = np.load(RUN_DIR / "data_evaluation" / "reconstructed_fields.npz")

for tag, vtu_path in DESIGNS.items():
    design = load_design(vtu_path, VF)
    sig_lo, sig_hi = solid_bounds(design, von_mises_stress(fields[f"{tag}__stress_0"]))
    eps_lo, eps_hi = solid_bounds(design, equivalent_strain(fields[f"{tag}__creep_strain_final"]),
                                  positive_only=True)
    print(f"{tag:<12s} initial sigma_eq: {sig_lo:.4f} to {sig_hi:.4f} MPa")
    print(f"{tag:<12s} final eps_eq:     {eps_lo:.6e} to {eps_hi:.6e}")
```
